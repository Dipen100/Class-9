#### <b>Data Types in Python</b>
>   Data types in python indicates what type of data is stored.
- Data Types used in python are:
    1. *Numeric Data Type*
    2. *Sequence Data Type*
    3. *Set Data Type*
    4. *Mapping Data Type*
    5. *Boolean Data Type*

##### 1) Numeric Data Type
- This type of data is used to store numbers.
- It helps us to do arithmetic operations.
- Different types of numeric data type are:
    1. Integer (int)
        - It may be whole number, positive, negative or zero.
        - It doesn't have any decimal or fraction value.
        - eg:- 0, -1, 2, -3, 4 etc.
    2. Floating-point (float)
        - It is used to represent numbers with decimal points.
        - It is used in scientific calculations or measurements.
        - eg:- 3.14, -7.5, 0.0 etc.
    3. Complex (complex)
        - It is used to represent complex number which has both real part and imaginary part.
        - They are commonly used in advanced mathematics, physics, engineering etc.
        - Format of complex number is: <b>a + bj</b>, where
            - a is the real part (floating or integer number).
            - b is the imaginary part (floating or integer number).
            - j is the imaginary unit (represents square root of -1).

```python
a = 10 # integer data type
b = 3.14 # floating data type
c = 3+4j # complex data type
print (type(c))
# print (type(b))
# print (type(c))
```

##### 2) Sequence Data Types
- It is used to store collection of items and items can be of any data type such as string or integer.
- In this data type, data are arranged in ordered.
- Different types of sequence data types are:
    1. String (str)
        >   String is a data type that is used to store string value or text.
        - It can include letters, numbers, symbols or spaces.
        - It is enclosed in single quotes (' ') or double quotes (" ")
        - eg:- "Hello", 'Python' etc.
    2. List (list)
        >   A list is an ordered collection of items which is mutable.
        - List is denoted with big brackets i.e [ ]
        - Mutable means we can change, add or remove items after it is created.
        - eg:- [1, 25, "apple"]
    3. Tuple (tuple)
        > A tuple is an ordered collection of items which is immutable.
        - Tuple is denoted with small brackets i.e ( )
        - Immutable means we cannot be changed after it is created.
        - eg:- (5, "banana", 7.8)

```python
string1 = "Virat Kohli"

list1 = [1,2,3,5] #list of integers
list2 = ["apple", "mango", "orange"] #list of strings

tuple1 = (1, 2, 3, "messi") #tuple with mixed items
```

##### 3) Set Data Types
> A set is a collection of unordered items with unique items.
- It is denoted by curly brackets i.e { }
- It doesn't allow duplicate values.
- We cannot access items using index.
- eg:- {1,2,3,4} ✅  {1,2,2,3} ❌

```python
set1 = {1,2,3,4}
set2 = {1,2,2,3}
print(f'value of set1: {set1}')
print(f'value of set2: {set2}')
```

##### 4) Mapping Data Type
- This data type is used to store data in key-value pairs.
1. Dictionary (dict)
    - It is a collection of data with key-value pairs.
    - Keys must be unique.

```python
'''
dict_name = {
    key : value
}
'''
# Example of Dictionary
student = {
    "name": "John",
    "age": 25,
    "address": "Butwal"
}
```

##### 4) Boolean Data Type (bool)
- The value of boolean data type can be either True or False.
- True = 1 and False = 0
- eg:- is_raining = True

```python
a = 10
b = 5
# Equality check
print(a == b) 

# Inequality check
print(a != b)   

# Greater than
print(a > b)  
```

#### **Operators in Python**
>   An operator is a symbol that tells the computer to perform certain task.
- There are various types of operators in python. Some are:
    - *Arithmetic Operators*
    - *Relational Operators*
    - *Logical Operators*
    - *Assignment Operators*


**1) Arithmetic Operators** <br>
>The operators which are used to perform mathematical calculations like addition, subtraction, multiplication, division, modulus and exponential are called arithmetic operators.
- eg:- 6 + 2, 5 * 2 <br> <br>

**2. Relational Operators** <br>
> The operators that are used to compare two values and return either true or false are called relational operators.
- eg:- 5>3, 2>=2 <br> <br>

**3. Logical Operators (AND, OR, NOT)** <br>
> It combines two or more relational expressions and return a single value as True or False.
- eg:- a AND b <br>
<img src="./images/logical-operators.png" height="200" width="500"/>

>*i) Logical AND* <br>
> The AND operator gives a true result only when both inputs are true otherwise gives false result. <br> <br>
ii)*Logical OR* <br>
> The OR operator gives a true result when any one or all input is true. <br> <br>
iii) *Logical NOT* <br>
> The NOT operator converts true result to false and false result to true.

```python
a = 5; b = 10; c = 15
if a > b and a > c: # both condition must be true
    print("a is larger")
elif b>a and b > c:
    print("b is larger")
else:
    print("c is larger")
```

##### **Operands**
> Operands are the variables or constant values on which the operation is performed.<br>
The expression is: a + b,
- '+' is an operator
- a and b are operand
- the whole expression is called operation

```python
# example of operand
a = 10; b = 5
sum = a + b # Here '+' is operator, a and b are operand
print (f'The sum of {a} and {b} is {sum}.')
```

##### **Expression**
> The combination of variables, operators, and constants that are evaluated to produce a result is called expression.
- The result of the expression is either string data, numeric data or logical data(true or false).
- eg:- SI = (P * T * R / 100) <br> <br>
    <img src="./images/expression.png" height="200" width="400"/>

```python
# Example of expression
P = 10000
T = 2
R = 1
SI = (P*T*R)/100
print(f'The Simple Interest of Rs{P} for {T}yrs and {R}% rate of interest is Rs {SI}')
```
